```python
temp = int(input())
# 50-75 : Extreme HOT
# 25-50 : HOT
# 10-24 : Cold
# <10 : Extremely Cold
if temp<=75 and temp>=50:
    print("Extreme Hot")
elif temp<=50 and temp>=25:
    print("Hot")
elif temp<=24 and temp>=10:
    print("Cold")
elif temp<10:
    print(" Extremely Cold")
```

```python
# Factorial of a number:
# 5! = 1*2*3*4*5

n=5
fact = 1
i = 1
while i<=n:
    fact=n*(n-1)
return fact
    
```
