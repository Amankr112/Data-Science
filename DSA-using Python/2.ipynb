#### Shallow and deep copy :

```python
list1 = [1,2,3]
list2 = list1  # Deep copy

list2.append(4)
print(list1)
print(list2)
```

```python
# Addresses of the list are same
list1 = [1,2,3]
list2 = list1  

list2.append(4)

print(id(list1))
print(id(list2))
```

```python
list1 = [1,2,3]
list2 = list1.copy()  # shallow copy

list2.append(4)
print(list1)
print(list2)
```

```python
# Addresses of the lists are different 
list1 = [1,2,3]
list2 = list1.copy()

list2.append(4)

print(id(list1))
print(id(list2))
```

#### Slicing :

```python
list1 = [1,2,3,5,6,8,9]
print(list1[2:6]) # Print values between index 2 to 5
print(list1[1:6:2]) # Print values between index 1 to 5, jumping 2
print(list1[6:1:-1]) # Print values between index 6 to 2 in reverse order
print(list1[:6]) # Print values between index 0 to 5
print(list1[2:]) # Print values between index 2 to end
print(list1[:]) # print all indexes of the list
```

```python
# Reverse a list using slicing :
list1 = [1,2,3,5,6,8,9]
print(list1[: : -1])
```

## Tuple () :

```python
tuple1 = ('hello', 123, 23.34, True, "aman")
print(tuple1)
print(len(tuple1))
print(type(tuple1))
```

```python
print(tuple1[2])
print(tuple1[-2])
```

```python
tuple1 = [1,2,3]
tuple2 = tuple1  

print(tuple1, id(tuple1))
print(tuple2, id(tuple1))
```

```python
tuple1 = (1,2,13,5,46,85,19,67,56,23)
print(tuple1[2:6])
print(tuple1[1:6:2])
print(tuple1[:6])
print(tuple1[2:])
print(tuple1[2:8:3])
print(tuple1[8:2:-2])
```

```python
# We can convert a tuple to list and vice-versa

tuple1 = (1,2,13,5,46,85,19,67,56,23)
list1 = list(tuple1)
print(list1)
print(type(list1))
```

```python
list1 = [1,2,13,5,46,85,19,67,56,23]
tuple1 = tuple(list1)
print(tuple1)
print(type(tuple1))
```
