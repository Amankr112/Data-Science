```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
dataset = pd.read_csv('heart.csv')
```

```python
dataset.head()
```

```python
dataset.shape
```

```python
dataset.describe()
```

```python
# Checking missing values
dataset.isnull().sum()
```

```python
# Distribution of Target Variable
dataset['target'].value_counts()
```

- 1 --> Defective Heart
- 0 --> Healthy Heart

#### Splitting the Features and Target

```python
X = dataset.drop(columns = 'target', axis = 1)
Y = dataset['target']
```

print(X)

```python
print(Y)
```

#### Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, stratify = Y, random_state = 3)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

#### Comparing the performance of the models

```python
# List of models
models = [LogisticRegression(max_iter = 1000), SVC(kernel = 'linear'), KNeighborsClassifier(), RandomForestClassifier()]
```

```python
def compare_models_train_test():
    
    for model in models:

        # training the model
        model.fit(X_train, Y_train)

        # Evaluating the model
        test_data_prediction = model.predict(X_test)
        
        accuracy = accuracy_score(Y_test, test_data_prediction)
        
        print('Accuracy score of the', model, 'is:', accuracy)
        
```

```python
compare_models_train_test()
```

---

### Cross Validation for different algorithms:

#### Using Logistic Regression :


```python
cv_score_lr = cross_val_score(LogisticRegression(max_iter = 10000), X, Y, cv=5)
```

```python
print(cv_score_lr)
```

##### Accuracy score for all 5 iterations

```python
# Mean accuracy 
mean_accuracy_lr = sum(cv_score_lr)/len(cv_score_lr)
mean_accuracy_lr = mean_accuracy_lr*100      # Converting to percentage
mean_accuracy_lr = round(mean_accuracy_lr, 2)   # only 2 decimal places will be there
print(mean_accuracy_lr)
```

#### Using support classifier:


```python
cv_score_svc = cross_val_score(SVC(kernel = 'linear'), X, Y, cv=5)
print(cv_score_svc)

# Mean accuracy 
mean_accuracy_svc = sum(cv_score_svc)/len(cv_score_svc)
mean_accuracy_svc = mean_accuracy_svc*100      # Converting to percentage
mean_accuracy_svc = round(mean_accuracy_svc, 2)   # only 2 decimal places will be there
print(mean_accuracy_svc)
```

#### Using KNN Classifier:

```python
cv_score_KNN = cross_val_score(KNeighborsClassifier(metric = 'manhattan'), X, Y, cv=5)
print(cv_score_KNN)

# Mean accuracy 
mean_accuracy_KNN = sum(cv_score_KNN)/len(cv_score_KNN)
mean_accuracy_KNN = mean_accuracy_KNN * 100      # Converting to percentage
mean_accuracy_KNN = round(mean_accuracy_KNN, 2)   # only 2 decimal places will be there
print(mean_accuracy_KNN)
```

#### Using Random Forest

```python
cv_score_RF = cross_val_score(RandomForestClassifier(random_state = 3), X, Y, cv=5)
print(cv_score_RF)

# Mean accuracy 
mean_accuracy_RF = sum(cv_score_RF)/len(cv_score_RF)
mean_accuracy_RF = round(mean_accuracy_RF * 100, 2)   
print(mean_accuracy_RF)
```

---

#### Creating a function to compare all models:


```python
# List of models
models = [LogisticRegression(max_iter = 1000), SVC(kernel = 'linear'), KNeighborsClassifier(), RandomForestClassifier()]
```

```python
def compare_models_cross_validation():
    for model in models:
        cv_score = cross_val_score(model, X, Y, cv=5)
        mean_accuracy = sum(cv_score)/len(cv_score)
        mean_accuracy = round(mean_accuracy*100, 2) 
        print('Cross Validation accuracy for the', model, '=  ', cv_score)
        print('Accuracy % of the', model, mean_accuracy)
        print('-----------------------------------------')
```

```python
compare_models_cross_validation()
```
