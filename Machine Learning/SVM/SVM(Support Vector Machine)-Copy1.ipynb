##### Gradient Descent is an optimization algorithm used for minimizing the loss function in various machine learning algorithms. It is used for updating the parameters of the learning model.

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import svm
```

```python
diabetes_data = pd.read_csv('diabetes.csv')
```

```python
diabetes_data.head()
```

```python
diabetes_data.shape
```

```python
# Getting the statistical measures of the dataset
diabetes_data.describe()
```

```python
diabetes_data['Outcome'].value_counts()
```

- 0 -> Non-Diabetic
- 1 -> Diabetic

```python
# Separating the features and target
features = diabetes_data.drop(columns ='Outcome', axis = 1) # axis=1 (to drop a column)
target =  diabetes_data['Outcome']                          # axis=0 (to drop a rows)
```

```python
print(features)
```

```python
print(target)
```

### Data Standardization:

```python
scaler = StandardScaler()
```

```python
scaler.fit(features)
```

```python
standardized_data = scaler.transform(features)
```

```python
print(standardized_data)
```

```python
# feature into the standardized_data formate 
features = standardized_data
target =  diabetes_data['Outcome']  
```

```python
print(features)
```

```python
print(target)
```

### Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(features, target, test_size=0.2, random_state=2)
```

```python
print(features.shape, X_train.shape, X_test.shape)
```

### Training the Model

```python
classifier = svm.SVM_classifier(learning_rate = 0.001, no_of_iterations = 1000, lambda_parameter = 0.01)
```

```python
# Training the svm classifier with the training data
classifier.fit(X_train, Y_train)
```

### Accuracy Score

```python
# Accuracy on training data
X_train_prediction = classifier.predict(X_train)
training_data_accuracy = accuracy_score(Y_train, X_train_prediction)
```

```python
print('Accuracy Score on training data =', training_data_accuracy)
```

```python
# Accuracy on test data
X_test_prediction = classifier.predict(X_test)
test_data_accuracy = accuracy_score(Y_test, X_test_prediction)
```

```python
print('Accuracy Score on test data =', test_data_accuracy)
```
