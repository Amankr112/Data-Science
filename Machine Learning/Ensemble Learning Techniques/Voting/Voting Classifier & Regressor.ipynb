```python
from sklearn.datasets import load_iris, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression, LinearRegression
from sklearn.ensemble import RandomForestClassifier, RandomForestRegressor, VotingClassifier, VotingRegressor

# --- Voting Classifier Example ---
X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

clf1 = LogisticRegression(max_iter=1000)
clf2 = RandomForestClassifier()

voting_clf = VotingClassifier(estimators=[('lr', clf1), ('rf', clf2)], voting='soft')
voting_clf.fit(X_train, y_train)
print("Voting Classifier Accuracy:", voting_clf.score(X_test, y_test)*100)

# --- Voting Regressor Example ---
X, y = load_diabetes(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=42)

reg1 = LinearRegression()
reg2 = RandomForestRegressor()

voting_reg = VotingRegressor(estimators=[('lr', reg1), ('rf', reg2)])
voting_reg.fit(X_train, y_train)
print("Voting Regressor R^2:", voting_reg.score(X_test, y_test)*100)
```
