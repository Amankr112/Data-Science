```python
# importing the dependencies
import numpy as np
import pandas as pd
import sklearn.datasets
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import RandomizedSearchCV
```

```python
# loding the datset from sklearn 
Breast_Cancer_dataset = sklearn.datasets.load_breast_cancer()
```

```python
print(Breast_Cancer_dataset)
```

```python
# Loading the data to dataframe
data_frame = pd.DataFrame(Breast_Cancer_dataset.data, columns = Breast_Cancer_dataset.feature_names)
```

```python
data_frame.head()
```

```python
data_frame.shape
```

```python
# Adding the 'target' column to the data frame
data_frame['label'] = Breast_Cancer_dataset.target
```

```python
data_frame.head()
```

```python
data_frame.shape
```

```python
# Checking the missing values
data_frame.isnull().sum()
```

```python
# Checking the distribution of target Variable
data_frame['label'].value_counts()
```

- 1--> Benign
- 0--> Malignant

#### Separating the features and target

```python
X = data_frame.drop(columns = 'label', axis=1)
Y = data_frame['label']
```

```python
print(X)
```

```python
print(Y)
```

```python
X = np.asarray(X)
Y = np.asarray(Y)
```

# **GridSearchCV**

- GridSearchCV is used for determining the best parameters for our model

```python
 # Loding the svm model
model = SVC()
```

```python
# Hyperparameters
parameters = {
    'kernel':['linear', 'poly', 'rbf', 'sigmoid'],
    'C':[1, 5, 10, 20]      # small value of C separates the classes effectively           
}
```

```python
# Grid Search
classifier = GridSearchCV(model, parameters, cv=5) 
```

```python
# Fitting the data to our model 
classifier.fit(X, Y)
```

```python
classifier.cv_results_
```

```python
# Best Parameters
best_parameters = classifier.best_params_
print(best_parameters)
```

```python
# Highest Accuracy
highest_accuracy = classifier.best_score_
print(highest_accuracy)
```

```python
# loading the results to pandas dataframe
result = pd.DataFrame(classifier.cv_results_)
```

```python
result.head()
```

```python
grid_search_result = result[['param_C', 'param_kernel', 'mean_test_score']]
```

grid_search_result

##### Highest Accuracy = 95.2%
##### Best Parameters = {'C':10, 'kernel':'linear'}

# **RandomizedSearchCV**

```python
# loading the SVC model
model = SVC()
```

```python
# hyperparameters

parameters = {
              'kernel':['linear','poly','rbf','sigmoid'],
              'C':[1, 5, 10, 20]
}
```

```python
# Randomized Search
classifier = RandomizedSearchCV(model, parameters, cv=5)
```

```python
# fitting the data to our model
classifier.fit(X, Y)
```

```python
classifier.cv_results_
```

```python
# Best Parameters
best_parameters = classifier.best_params_
print(best_parameters)
```

```python
# higest accuracy

highest_accuracy = classifier.best_score_
print(highest_accuracy)
```

```python
# loading the results to pandas dataframe
result = pd.DataFrame(classifier.cv_results_)
```

```python
result.head()
```

```python
randomized_search_result = result[['param_C', 'param_kernel', 'mean_test_score']]
```

```python
randomized_search_result
```

Highest Accuracy = 95.2%

Best Parameters = {'C':10, 'kernel':'linear'}
