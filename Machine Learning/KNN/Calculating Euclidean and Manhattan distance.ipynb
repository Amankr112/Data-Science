```python
import numpy as np
import math
```

```python
# Considering two points with 2 dimensions 
p1 = (1,1)
p2 = (2,2)
```

#### Calculating Euclidean Distance :

```python

dist = (p1[0]-p2[0])**2 +  (p1[1]-p2[1])**2
Euclidean_Distance = np.sqrt(dist)
print(Euclidean_Distance)
```

```python
# Considering two points with 3 dimensions 
p1 = (1,1,1)
p2 = (2,2,2)

dist = (p1[0]-p2[0])**2 + (p1[1]-p2[1])**2 + (p1[2]-p2[2])**2
Euclidean_Distance = np.sqrt(dist)
print(Euclidean_Distance)
```

```python
# Considering two points with 4 dimensions 
p1 = (1,1,1,1)
p2 = (2,2,2,2)

dist = 0
for i in range (len(p1)):
    dist = dist + (p1[i]-p2[i])**2

Euclidean_Distance = np.sqrt(dist)
print(Euclidean_Distance)
```

#### Creating function for calculating the Euclidean Distance :

```python
def get_euclidean_distance(p1, p2):
    dist = 0
    for i in range (len(p1)):
        dist = dist + (p1[i]-p2[i])**2
    
    Euclidean_Distance = np.sqrt(dist)
    print(Euclidean_Distance)
    
```

```python
get_euclidean_distance((1,1,1,1), (2,2,2,2))
```

```python
get_euclidean_distance((1,1,1), (2,2,2))
```

```python
get_euclidean_distance((1,1), (2,2))
```

---

#### Calculating Manhattan Distance :


```python
def get_manhattan_distance(p1, p2):
    
    dist = 0
    for i in range (len(p1)):
        dist = dist + abs(p1[i] - p2[i])    # abs -> convert negative no. to positive no.
        manhattan_distance = dist
    
    print(manhattan_distance)
    
```

```python
get_manhattan_distance((1,1,1,1), (2,2,2,2))
```

```python
get_manhattan_distance((1,1,1), (2,2,2))
```

```python
get_manhattan_distance((1,1), (2,2))
```

#### Calculating distance between two data points :

```python
# Distance between two similar data points - less

print(get_euclidean_distance((6,148,72,35,0,33.6,0.627,50),(8,183,64,0,0,23.3,0.672,32)))
print(get_manhattan_distance((6,148,72,35,0,33.6,0.627,50),(8,183,64,0,0,23.3,0.672,32)))
```

```python
# Distance between two dissimilar data points - more

print(get_euclidean_distance((6,148,72,35,0,33.6,0.627,50),(1,89,66,23,94,28.1,0.167,21)))
print(get_manhattan_distance((6,148,72,35,0,33.6,0.627,50),(1,89,66,23,94,28.1,0.167,21)))
```
