**K-Nearest Neighbors Classifier:**

K-Nearest Neighbors (KNN) is a non-parametric, supervised machine learning algorithm used for both classification and regression tasks. It is considered a "lazy learner" because it does not build a model during a training phase; instead, it stores the entire training dataset and performs computations only when a prediction is required for a new data point.

```python
class KNN_classifier():

     # Initiating the parameters
    def __init__(self, distance_metric):
        self.distance_metric = distance_metric

    # getting the distance metric
    def get_distance_metric(self, training_data_point, test_data_point):
        if (self.distance_metric == 'euclidean'):
            dist = 0
            for i in range (len(training_data_point) - 1):    # Excluding the target column
                dist += (training_data_point[i] - test_data_point[i])**2

            euclidean_dist = np.sqrt(dist)
            return euclidean_dist
            
        elif (self.distance_metric == 'manhattan'):
            dist = 0
            for i in range (len(training_data_point) - 1):
                dist += abs(training_data_point[i] - test_data_point[i])
                manhattan_dist = dist
                return manhattan_dist
        
                

    # getting the nearest neighbors
    def nearest_neighbor(self, X_train, test_data, k):
        distance_list = []
        for training_data in X_train:
            distance = self.get_distance_metric(training_data, test_data)
            distance_list.append((training_data, distance))

        distance_list.sort(key=lambda  x:x[1])  # x[1] means we are taking distance and sorting them
        
        neighbors_list = []

        for j in range(k):
            neighbors_list.append(distance_list[j][0])   # here j representing the training data & we don't need distance here so it is 0.
        return neighbors_list

    
    # Predict the class of the new data points:
    def predict(self, X_train, test_data, k):
        neighbors = self.nearest_neighbor(X_train, test_data, k)

        for data in neighbors:
            label = []
            label.append(data[-1])  # data[-1] represent values of outcome
        predicted_class = statistics.mode(label)   # mode will find the most frequent class to decide in which class the points will go.
        return predicted_class
        
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
dataset = pd.read_csv('diabetes.csv')
```

```python
# Separating the features and Target
X = dataset.drop(columns = 'Outcome', axis = 1)
Y = dataset['Outcome']
```

```python
# converting the data to numpy array:
X = X.to_numpy()
Y = Y.to_numpy()
```

### Train Test Split

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, stratify = Y, random_state = 2)
```

- X_train --> Training data with features and target
- Y_train --> Test data without target

### Model Training 

##### Using euclidean distance

```python
classifier = KNN_classifier(distance_metric ='euclidean')
```

##### The KNN classifier can predict the label for only one data point at a time (one target column)

```python
prediction = classifier.predict(X_train, X_test[0], k=5)
```

#### Model Evaluation :
