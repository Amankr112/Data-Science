```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
```

```python
def analyzer(max_depth):
    data = pd.read_csv('Social_Network_Ads.csv')

    X = data.iloc[:, [0, 1]].values   # Age & EstimatedSalary
    y = data.iloc[:, 2].values        # Purchased

    clf = DecisionTreeClassifier(max_depth=max_depth)
    clf.fit(X, y)

    # use reasonable steps
    a = np.arange(start=X[:, 0].min()-1, stop=X[:, 0].max()+1, step=1)
    b = np.arange(start=X[:, 1].min()-1000, stop=X[:, 1].max()+1000, step=1000)

    XX, YY = np.meshgrid(a, b)

    input_array = np.c_[XX.ravel(), YY.ravel()]
    labels = clf.predict(input_array)

    plt.contourf(XX, YY, labels.reshape(XX.shape), alpha=0.5, cmap="coolwarm")
    plt.scatter(X[:, 0], X[:, 1], c=y, edgecolor="k")
    plt.xlabel("Age")
    plt.ylabel("Estimated Salary")
    plt.show()
```

```python
analyzer(max_depth = None)
```

```python
analyzer(max_depth = 1)
```

```python
analyzer(max_depth = 2)
```
