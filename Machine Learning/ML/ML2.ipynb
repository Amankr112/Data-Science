### Linear Regression: 
It is a supervised algorithm that is used to find out the dependent variable with the help of independent variableand by drawing best fit line.


```python
#### Best Fit line : 
A straight line from where the distance of the data point will be minimum.
    
#### cost function : 
It is used to calculate best slope and intercept value.
    
#### Gradient Descent : 
If we join the data point in slope vs cost polt then the curve line that we get is called as gradient descent curve.

#### Global Minimum : 
The point at gradient descent curve, when cost become minimum and accuracy is highest.

#### learning rate : 
- It is used to calucalte how fast the model can learn, Generally alpha = 0.001
- If we take larger value of alpha then it may miss the gobale minima.
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, confusion_matrix
import pandas as pd
import numpy as np
```

```python
d = fetch_california_housing()
d
```

```python
df = pd.DataFrame(d.data, columns = d.feature_names)
```

```python
df
```

```python
df['Price'] = d.target
```

```python
df
```

```python
X = df.drop('Price', axis=1)
Y = df['Price']
```

```python
X
```

```python
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 41)
```

```python
LR = LinearRegression()
```

```python
# Training the dataset
LR.fit(X_train, Y_train)
```

```python
pred = LR.predict(X_test)
pred
```

```python
r2 = r2_score(Y_test, pred)
r2
```

```python
MAE = mean_absolute_error(Y_test, pred)
MAE
```

```python
MSE = mean_squared_error(Y_test, pred)
MSE
```

### R-2 Square :
-> It is used to evaluate the accuracy of the model by calculating the residual.
-> Formula :   
              =  1 - (sum squared of residual/sum squared of Actual)
              Yi = Actual Value
              Yi^ = Predicted Value
              Yi bar = Mean of Actual Value
                
-> Limitations :  

- If the difference between the actual o/p and predicted o/p is very less, then the r-2 score value can be -ve.
- If we Continously add new features then r-2 score value will keep on increasing without thinking if the column is
   co-related with the o/p feature or not.
