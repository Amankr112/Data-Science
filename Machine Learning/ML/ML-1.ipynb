## label Encoding :

```python
from sklearn.preprocessing import LabelEncoder   
import pandas as pd
```

```python
df = pd.DataFrame({'Climate': ['sunny', 'Rainy', 'windy', 'cloudy', 'rainy', 'stormy']})
```

```python
df
```

```python
label = LabelEncoder()
df['Encoded_Climate'] = label.fit_transform(df['Climate'])
df
```

```python
# Rainy will have the encoding as 0 because it has the lowest ASCII value.
```

## One hot Encoding :

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
d = {'Weather': ['Sunny', 'Rainy', 'Wnidy', 'cloudy', 'rainy', 'stormy']}
df = pd.DataFrame(d)
df
```

```python
# Created an object to access its properties:
OHE = OneHotEncoder()
OHE_Encoded = OHE.fit_transform(df['Weather'])
OHE_Encoded
```

```python
# It will throw an error because OHE requires 2-D value as input.(means 2 big bractes should be there)
```

```python
OHE = OneHotEncoder()
OHE_Encoded = OHE.fit_transform(df[['Weather']])
OHE_Encoded
```

```python
values = OHE_Encoded.toarray()
```

```python
columns = OHE.get_feature_names_out()
```

```python
# Now we have to merge these values to their respectyed columns
df1 = pd.DataFrame(values, columns = columns)
df1
```
