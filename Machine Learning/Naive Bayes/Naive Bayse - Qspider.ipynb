```python
import pandas as pd
from sklearn.naive_bayes import GaussianNB
from sklearn.datasets import load_iris, load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
```

```python
d = load_iris()
d
```

```python
df = pd.DataFrame(d.data, columns = d.feature_names)
df
```

```python
df['variety'] = d.target
df
```

```python
X = df.drop('variety', axis=1)
X
```

```python
Y = df['variety']
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42)
```

```python
model = GaussianNB()
model.fit(X_train, Y_train)
```

```python
model.score(X_train, Y_train)
```

```python
model.score(X_test, Y_test)
```

```python
p = model.predict(X_test)
p
```

```python
confusion_matrix(Y_test, p)
```

```python
accuracy_score(Y_test, p)
```

```python
classification_report(Y_test, p)
```

```python
print(classification_report(Y_test, p))
```

```python
# For the diabetes dataset we will train Naive Bayes:
```

```python
a = load_diabetes()
a
```

```python
df1 = pd.DataFrame(a.data, columns = a.feature_names)
df1
```

```python
df1['output'] = a.target
df1
```

```python
# converting the floating value of output column into 0 and 1:
l = [1 if i>150 else 0 for i in df1['output']]
l
```

```python
X = df1.drop('output', axis=1)
X
```

```python
Y = l
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size = 0.3, random_state = 42)
```

```python
model1 = GaussianNB()
model1.fit(X_train, Y_train)
```

```python
model1.score(X_train, Y_train)
```

```python
model1.score(X_test, Y_test)
```

```python
p = model1.predict(X_test)
```

```python
p
```

```python
confusion_matrix(Y_test, p)
```

```python
print(classification_report(Y_test, p))
```
