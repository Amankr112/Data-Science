```python
# WAP to reverse a string using while loop:
string = input("Enter a string:")
i = 0
reverse =''
while(i<len(string)):
    reverse = string[i] + reverse
    i+=1
print(reverse)
```

```python
# WAP to add all the digits inside the given number: 
# i/p = 245, o/p = 11

n = int(input('Enter a number:'))
sum_of_digits = 0
while(n>0):
    r = n%10
    sum_of_digits += r
    n //= 10
print(sum_of_digits)
    
```

```python
# WAP to print the fibonacci series up to a given length:


n = int(input("Enter the length of the fibonacci series:"))
a, b = 0, 1

for i in range(n):
    print(a, end=' ')
    a, b = b, a+b

```

```python
# WAP to reverse alternate word of a given string:
#  i/p : we love to code in python
#  o/p : ew love ot code si python

s = input("Enter a string: ")
l = s.split()
res = ''
for i in range(len(l)):
    if i % 2 == 0:
        res += ' ' + l[i][::-1]
    else:
        res += ' ' + l[i]
print(res)
```

```python
# WAP to count the frequency of each word from a given string:
# i/p : mohan rohan mohan rohan sohan
#  o/p : mohan appeared 2 times
#        rohan appeared 2 times
#        sohan appread 1 time.


s = input("Enter a string: ")
words = s.split()
d = {}

for i in words:
    if i in d:
        d[i] += 1
    else:
        d[i] = 1
print(d)
```
