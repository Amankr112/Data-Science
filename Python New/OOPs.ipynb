```python
# __doc__ :  It is a predefined variable used to display/store documentation string.
class student():
    """ This class is used to store the information about student """
    def __init__(self, x, y):
        self.name = x
        self.roll = y

    def display(self):
        print('My name is ', self.name)
        print('My roll is ', self.roll)

s1 = student('Aman', 7)
s2 = student('Rahul', 8)
print(student.__doc__)
s1.display()
s2.display()    
```

```python
# Instance, local and static variable :
class Student():
    Institute_name = 'Qspider'   # Static var.
    def __init__(self, x, y):    # Here x, y is local var.
        self.name = x        # Here name and roll is Instance var.
        self.roll = y
    def display(self):
        print('My name is ', self.name)
        print('My roll is ', self.roll)
        print('My Institute name is ', self.Institute_name)
s1 = Student('Aman', 7)
s1.display()
```

```python
# Instance method : It is a type of method in which we access instance variable as well as static variable.
#                   If first argument of any method is self then that is a instance method.
class student():
    def __init__(self, x, y):
        self.name = x
        self.roll = y

    def display(self):    # Instance method
        print('My name is', self.name)
        print('My roll is', self.roll)

s1 = student('Raj', 1)
s1.display()
```

```python
# Class Method : It is a type of method in which we access static variable also.
#                To make a method as class method we have to use reference variable 
#                that is class as well as we have to write down the decorator -->  @classmethod.
class Student():
    Institute_name = 'Qspider'
    def __init__(self, x, y):
        self.name = x
        self.roll = y
    def display(self):
        print('My name is', self.name)
        print('My roll is', self.roll)
        print('My Institute name is ', self.Institute_name)

    @classmethod      # decorator
    def InstituteInfo(cls):    # class method
        print('My Institute name is', cls.Institute_name)

s1 = Student('Aman', 7)
s1.display()
s1.InstituteInfo()
```

```python
# Static method : It is a type of method which is used to add an extra functionality which is not related to instance as well as static variable.
#                 Inside the static method we should not used (self, cls) as argument.
#                 To make a method as static method we need a decorator -->  @staticmethod
class Student():
    Institute_name = 'Qspider'
    def __init__(self, x, y):
        self.name = x
        self.roll = y
    def display(self):
        print('My name is ', self.name)
        print('My roll is ', self.roll)
        print('My Institute name is ', self.Institute_name)

    @classmethod      # decorator
    def InstituteInfo(cls):    # class method
        print('My Institute name is', cls.Institute_name)
    
    @staticmethod    # decorator
    def game():
        n = int(input('Enter a number:'))
        for i in range(1, n+1):
            print(i)
            
s1 = Student('Aman', 7)
s2 = Student('Raj', 1)
s1.display()
s2.display()
s1.InstituteInfo()
s1.game()
s2.InstituteInfo()
s2.game()
```

#### Inheritance : 
Inheritance is a concept with the help of which whatever properties or functionalities belongs to the parent class directly belongs to the child class.

```python
# Hierarchical Inheritance:
class A():
    def car(self):
        print('Honda City')
class B(A):
    def bike1(self):
        print('Splender')
class C(B):
    def bike2(self):
        print('Pulser')

b = B()
b.car()
b.bike1()

d = C()
d.car()
d.bike2()
```

```python
# Method Overloading :  (Python does not support it)
class A():
    def m1(self):
        print('First method')
    def m1(self):
        print('Second method')
    def m1(self):
        print('Third method')

a = A()
a.m1()
```

```python
# Constructor Overloading : (Python doesn't support it)
class A():
    def __init__(self):
        print('First constructor')
    def __init__(self):
        print('Second constructor')
    def __init__(self):
        print('Third constructor')
a = A()
```

```python
# Operator Overloading : (Python supports it)  --> 

class book():
    def __init__(self, x):
        self.pages = x
    def __add__(self, other):
        return self.pages + other.pages
b1 = book(200)
b2 = book(300)
print(b1+b2)
```

```python
# Method Overriding : The concept with the help of which if child class is unhappy with the parent method 

class A():
    def car(self):
        print('Alto-800')

class B(A):
    def car(self):
        print('Scorpio')

b = B()
b.car()
```

```python
# Constructor Overriding : 

class A():
    def __init__(self):
        print('Parent constructor')

class B(A):
    def __init__(self):
        print('Child constructor')

    def __init__(self):
        print('baby constructor')

b = B()
```

```python
# Super: Using super we can access parent class instance method.

class A():
    def car(self):
        print('Alto-800')

class B(A):
    def car(self):
        print('Venue')
    def car2(self):
        super().car()

b = B()
b.car()
b.car2()
```

```python
# Using super builtin fuction we can access parent class constructor.

class A():
    def __init__(self):
        print('Parent constructor')

class B(A):
    def __init__(self):
        print('Child constructor')
        super().__init__()
b = B()
```

```python
# Using super we can access Static variable of a parent class using Super().
class A():
    x = 10

class B(A):
    x = 20
    def display(self):
        print("Value of x is", self.x)
        print('Value of parent x is', super().x)
b = B()
b.display()
```

```python
# How to modify the Instance Variable from outside the class.
class Student():
    def __init__(self, x, y):
        self.name = x
        self.age = y
    def display(self):
        print('My name is', self.name)
        print('My age is', self.age)

s1 = Student('Aman', 22)
s1.display()
s1.name = 'Chaman'
s1.display()
```

```python
# How to modify the Instance Variable from inside the class.

class Student():
    def __init__(self, x, y):
        self.name = x
        self.age = y

    def Name_Modification(self, p, q):
        self.name = p
        self.age = q

    def display(self):
        print('My name is', self.name)
        print('My age is', self.age)

s1 = Student('Aman', 24)
s1.display()
s1.Name_Modification('Raj', 20)
s1.display()
```

```python
# How to modify the Static Variable from inside the class.
class Student():
    school = 'xyz'
    @classmethod
    def change_school_name(cls, x):
        cls.school = x
        print('The school name is', cls.school)

s1 = Student()
print(s1.school)
s1.change_school_name('abc')
```

```python
# Public, Private, Protected.

# Public : If a member(variable/method) is public then we can access that member inside and outside the class.

class A():
    def __init__(self):
        self.x = 10
    def display(self):
        print('The value of x is:', self.x)
a = A()
a.display()
print(a.x)
```

```python
# Private : If a memeber function is private then we can access that member only inside the class.

class A():
    def __init__(self):
        self.__x = 10

    def display(self):
        print('The value of x is:', self.__x)
        
a = A()
a.display()
# print(a.x)   --> It will give error
```

```python
# Protected : If a member is protected then that member inside the parent class can be accessed only inside the parent class, 
#             but if we try to access protected member inside the child class then it can be accessed inside as well as outside the class.

# But it is like public as Protected is not applicable in python.

class A():
    def __init__(self):
        self._x = 7
    def display(self):
        print('Value of x is:', self._x)

class B(A):
    def display2(self):
        print('The Value of x is:', self._x)

b = B()
a = A()
b.display2()
print(b._x)
a.display()
print(a._x)
```

```python
# Getter & Setter :

# getter : It is a method with the help of which we are able to return or Extract the value of a instance variable.

class A():
    def __init__(self, x, y):
        self.name = x
        self.age = y
        
    def get_name(self):
        return self.name

    def get_age(self):
        return self.age

a = A('Aman', 22)
print(a.get_name())  
print(a.get_age())
    
```

```python
# Setter : It is a type of method which is used to update or modify the value of a instance variable.

class A():
    def __init__(self, x):
        self.name = x

    def get_name(self):
        return self.name

    def set_name(self, y):
        self.name = y
        
a = A('Aman')
print(a.get_name())
print(a.set_name('Raj'))   
print(a.get_name())
```

```python
# Encapsulation : It is a process of grouping or binding variables and methods inside a single unit. mainly used for data hiding / data safety.

class Bank():
    def __init__(self, x):
        self.__balance = x
        
    def get_balance(self, pin):
        if pin == 8421:
            return self.__balance

    def deposite(self, y):
        if y > 0:
            self.__balance = self.__balance + y

b = Bank(50000)
print(b.get_balance(8421))
b.deposite(10000)
print(b.get_balance(8421))

```

```python
# Abstraction : The process of highlighting the set of features and hiding the internal implementation of the code.

from abc import ABC, abstractmethod
class A(ABC):
    @abstractmethod
    def pay_loan(self):
        pass
    def details(self):
        print('The loan is paid')

class B(A):
    def pay_loan(self):
        print('The Home loan is paid')

class D(A):
    def pay_loan(self):
        print('The Car loan is paid')

b = B()
b.pay_loan()
b.details()
d = D()
d.pay_loan()
d.details()

# a = A()    --> we can't create the object of class which contains incomplete method.
```
