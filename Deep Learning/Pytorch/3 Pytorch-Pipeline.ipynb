```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv")
```

```python
X = df.drop(columns = ['diagnosis'], axis=1)
Y = df['diagnosis']
```

```python
# Train Test Split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# Scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
# We will do label encoding as the data in Y_train is not in numbers:
LE = LabelEncoder()
Y_train = LE.fit_transform(Y_train)
Y_test = LE.transform(Y_test)
```

```python
Y_train
```

```python
# Converting Numpy arrays to Pytorch Tensors:

X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
Y_train_tensor = torch.from_numpy(Y_train)
Y_test_tensor = torch.from_numpy(Y_test)
```

```python
# Defining the model :

class BreastCancerModel():

  def __init__(self, X):

    self.weights = torch.rand(X.shape[1], 1, dtype = torch.float64, requires_grad = True)
    self.bias = torch.zeros(1, dtype = torch.float64, requires_grad = True)

  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):
    # lamp prediction to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(Y_train_tensor * torch.log(y_pred) + (1 - Y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss
```

```python
# Parameters :
learning_rate = 0.1
epochs = 25
```

```python
# Training Pipeline :

# Creating Model for my training data -
model = BreastCancerModel(X_train_tensor)
```

```python
model.weights
```

```python
model.bias
```

```python
# Training the model:

# Define Loop -
for epoch in range(epochs):

  # forward pass
  y_pred = model.forward(X_train_tensor)

  # loss calculation
  loss = model.loss_function(y_pred, Y_train_tensor)

  # backward pass
  loss.backward()

  # parameters update
  with torch.no_grad():
    model.weights.data = model.weights.data - learning_rate * model.weights.grad
    model.bias.data = model.bias.data - learning_rate * model.bias.grad

  # Zero gradient
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  # print loss in each epoch :
  print(f'Epoch : {epoch + 1}, Loss : {loss.item()}')
```

```python
# Model Evaluation :
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.9).float()

  accuracy = (y_pred == Y_test_tensor).float().mean()
  print(f'Accuracy : {accuracy.item()}')
```
