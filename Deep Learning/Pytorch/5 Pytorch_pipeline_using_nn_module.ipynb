```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv")
```

```python
df.drop(columns = ['id', 'Unnamed: 32'], inplace = True, axis=1)
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
X = df.drop(columns = ['diagnosis'], axis=1)
Y = df['diagnosis']
```

```python
# Train Test Split
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)
```

```python
# Scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
Y_train
```

```python
# We will do label encoding as the data in Y_train is not in numbers:
LE = LabelEncoder()
Y_train = LE.fit_transform(Y_train)
Y_test = LE.transform(Y_test)
```

```python
# Converting Numpy arrays to Pytorch Tensors:

X_train_tensor = torch.from_numpy(X_train).float()
X_test_tensor = torch.from_numpy(X_test).float()

Y_train_tensor = torch.from_numpy(Y_train).float()
Y_test_tensor = torch.from_numpy(Y_test).float()
```

```python
# Defining the model :
import torch.nn as nn
class BreastCancerModel(nn.Module):

    def __init__(self, num_features):
      super().__init__()
      self.linear = nn.Linear(num_features, 1)
      self.sigmoid = nn.Sigmoid()

    def forward(self, features):
        out = self.linear(features)
        out = self.sigmoid(out)
        return out
```

```python
# Parameters :
learning_rate = 0.1
epochs = 25
```

```python
# Defining loss function :
loss_function = nn.BCELoss()
```

```python
# Training Pipeline :

# Creating Model for my training data -
model = BreastCancerModel(X_train_tensor.shape[1])

# Define optimizer:
optimizer = torch.optim.SGD(model.parameters(), lr = learning_rate)

# define loop
for epoch in range(epochs):

    # forward pass
    y_pred = model(X_train_tensor)

    # loss calculate
    loss = loss_function(y_pred, Y_train_tensor.reshape(-1,1))

    # Zero gradient (Clear gradients)
    optimizer.zero_grad()

    # backward pass
    loss.backward()

    # parameters update
    optimizer.step()

    # print loss in each epoch :
    print(f'Epoch : {epoch + 1}, Loss : {loss.item()}')
```

```python
# Model Evaluation :
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.5).float()

  accuracy = (y_pred == Y_test_tensor).float().mean()
  print(f'Accuracy : {accuracy.item()}')
```
