#### Difference between Series and Array.

#### Series -
- It belongs to pandas.
- complex ke baad everything is considered as object
- It displays data, index and datatype
- pd.Series(collection)


#### Array -
- It belongs to numpy
- string ke baad everything is considered as object
- It displays only data
- np.array(collection)

#### Difference Between Array and DataFrame

| **Array** | **DataFrame** |
|------------|---------------|
| An array is a collection of elements of the **same data type**. | A DataFrame is a **2-dimensional tabular structure** with rows and columns. |
| Provided by libraries like **NumPy**. | Provided by the **Pandas** library. |
| Usually stores **homogeneous** data (same type). | Can store **heterogeneous** data (different data types in different columns). |
| Uses **numeric indexing** only. | Uses both **row and column labels** (index and column names). |
| Faster and more memory-efficient for numerical computations. | Easier for data analysis, cleaning, filtering, and manipulation. |
| Does not have column names. | Has column names and row indexes. |
| Mainly used for mathematical and scientific computations. | Mainly used for data analysis, machine learning, and data preprocessing. |

```python
# Filling Nan values :
1) fillna() : 
2) ffill() :
3) bfill() :

1) fillna() : It will fill the 'nan' value with specific value
2) ffill() :
3) bfill() :
```

<!-- Filling Nan values:
1) Fillna()
2) ffill()
3) bfill()

1. fillna() : It will fill the "Nan" value with specified value.

    syntax :        var['col_name'] = var['col_name'].fillna('value')
    

2. ffill() : It will fill the value with the previous value.

    syntax :        var['col_name'] = var['col_name'].ffillna() 


3. bfill() : It will fill the value with the next value.

    syntax :        var['col_name'] = var['col_name'].bfillna() 


-->
