#### Logical Operators :
1.) AND (&)

2.) OR (|)

3.) NOT (~)

```python
# Find out the ages which are greater than 18 but less than 60.

age[(age>18) & (age<60)]
```

```python
# Filter out the values which are divisible by 5 or 7 in 'b'.
b[(b%5==0) | (b%7==0)]
```

```python
c
```

```python
# Find out the ages which are less than 18 and greater than 60.

age[(age < 18) | (age > 60)]
```

```python
# Find out the ages which are between 18 and 60.
age[~((age < 18) | (age > 60))]
```
