```python
# Day - 2

boolean_data = pd.read_csv('https://github.com/Taj1920/Data-Analysis-Datasets/raw/refs/heads/main/Datasets-Practice/sales_data_boolean_filtering.csv')
boolean_data
```

```python
res = boolean_data.groupby('Region')['TotalSales'].sum()
res = pd.DataFrame(res)
res.reset_index(inplace = True)
res
```

```python

plt.figure(figsize = (15, 5))
plt.plot(res['Region'], res['TotalSales'], label = 'Region', marker = 'o', color = 'red', markerfacecolor = 'Blue', linewidth = 2, linestyle = 'solid')
plt.legend(title = 'Total-Sales', fontsize = 15, loc = 'upper right')
plt.xlabel('Region', fontsize = 15, color = 'blue')
plt.ylabel('Total-Sales', fontsize = 15, color = 'blue')
plt.title('Total-Sales-Graph', fontsize = 10, color = 'Blue')
plt.grid(True)
for i in range(len(res)):
    plt.annotate(f"({res['Region'][i]}, {res['TotalSales'][i]})", (res['Region'][i], res['TotalSales'][i]), ha = 'center')
```

```python
x = np.arange(1, 11)
z = x**3
plt.figure(figsize = (15, 5))
plt.plot(x,z, label = 'X-Values', marker = 'o', color = 'red', markerfacecolor = 'Blue', linewidth = 2, linestyle = 'dotted')
plt.legend(title = 'Values', shadow = True, fontsize = 15, loc = 'upper center')
plt.xlabel('values 1 to 10' , fontsize = 10, color = 'blue')
plt.ylabel('values 0 to 1000', fontsize = 10, color = 'blue')
plt.title('Total-Sales-Graph', fontsize = 15, color = 'Blue')
plt.grid(True)
for i in range(len(x)):
    plt.annotate(f"({x[i]}, {z[i]})", (x[i], z[i]), ha = 'center')
```
