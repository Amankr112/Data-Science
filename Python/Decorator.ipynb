```python
# Closures : A nested function that remembers and uses variables from its enclosing scope even after the outer function has finished execution.
def main_welcome():
    msg="Hey I am Aman"
    def sub_welcome_class():
        print("Welcome to my world")
        print(msg)
        print("Please come to my home")
    return sub_welcome_class()
```

```python
main_welcome()
```

```python
# Closures
def main_welcome(msg):
    def sub_welcome_class():
        print("Welcome to my world")
        print(msg)
        print("Please come to my home")
    return sub_welcome_class()

#all the variables defined in main function can be accessible in sub function
```

```python
# Closures or Initial Decorators
def main_welcome(func):
    def sub_welcome_class():
        print("Welcome to my world")
        print(func([1,2,3,4,5]))
        print("Please come to my home")
    return sub_welcome_class()
```

```python
main_welcome(len)
```

```python
# Decorator: A function that wraps another function to modify or extend its behavior without changing its source code.
def main_welcome(func):
    def sub_welcome_class():
        print("Welcome to my world")
        func()
        print("Please come to my home")
    return sub_welcome_class()
```

```python
def channel_name():
    print("This is Aman's youtube channel")
```

```python
main_welcome(channel_name)
```

```python
@main_welcome    #decorator name
def channel_name():
    print("This is Aman's youtube channel")
```
