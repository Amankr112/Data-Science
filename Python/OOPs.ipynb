### Inheritance
##### Inheritance is an object-oriented programming (OOP) feature that allows one class (child/derived class) to inherit properties and methods from another class (parent/base class).
- Code reuse: Avoid rewriting the same code.
- Extensibility: Easily extend or modify functionality.
- Organization: Helps structure complex programs better.



```python
class car:
    def __init__(self, window, door, engine_type):
        self.windows=window
        self.doors=door
        self.enginetype=engine_type
    def drive():
        print("The person drives the car")
```

```python
car1=car(6, 5, "Petrol")
```

```python
car1.windows
```

```python
car.drive()
```

```python
class audi(car):
    def __init__(self, window, door, engine_type, enableAI):
        super().__init__(window, door, engine_type)
        self.enableai=enableAI
    def selfdriving(self):
        print("Audi supports self driving")
```

```python
AudiQ7=audi(4,4,"Desiel",True)
```

```python
AudiQ7.enableai
```

```python
type(AudiQ7.enableai)
```

```python
audi.drive()
```

```python
AudiQ7.selfdriving()
```

### Class Methods and Class Variables 

#### Class Variables:
- Definition: Variables that are shared across all instances of a class.
- Stored in the class and not in individual instances.
- Accessed using: ClassName.variable or self.__class__.variable

#### Class Methods:
- Definition: Methods that are bound to the class and not the instance.
- Use the @classmethod decorator.
- Take cls (not self) as the first parameter to refer to the class.
- Often used to manipulate class variables or create alternative constructors.

```python
class Car:
    base_price=500000   #class variable
    def __init__(self, windows, seats, doors, power):
        self.window=windows
        self.seat=seats
        self.door=doors
        self.power=power
    def what_is_base_price(self):
        print("The base Price is {}".format(self.base_price))
```

```python
car1=Car(4,5,4,200000)
```

```python
car1.base_price
```

```python
Car(4,7,5,300000)
```

```python
class Car:
    base_price=500000   #class variable
    def __init__(self, windows, seats, doors, power):
        self.window=windows
        self.seat=seats
        self.door=doors
        self.power=power
    def what_is_base_price(self):
        print("The base Price is {}".format(self.base_price))

    @classmethod      #Decorator
    def revised_Base_Price(cls, inflation):
        cls.base_price=cls.base_price + cls.base_price * inflation
```

```python
Car.revised_Base_Price(.10)
```

```python
Car.base_price
```

```python
car1=Car(4,7,5,300000)
```

```python
car1.revised_Base_Price(0.10)
```

```python
car1.base_price
```
