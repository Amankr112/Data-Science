```python
# Prime
def Prime_no(n):
    if n<1:
        print('Enter a valid number')
    else:
        for i in range(2, n):
            if n%i == 0:
                print('Not a Prime Number')
                break
            else:
                print('Prime Number')
                break
n = int(input('Enter a number:'))
Prime_no(n)
```

```python
# Strong Number

n = int(input('Enter a number:'))
temp = n
total = 0
while temp>0:
    rem = temp%10
    fact = 1
    for i in range(1, rem+1):
        fact = fact*i
    total = total+fact
    temp //=10

if total == n:
    print('Strong Number')
else:
    print('Not a strong Number')
```

```python
# WAP to get the following output.
# In = 'push maadi kushi padi'
# o/p = {'push':'ph', 'maadi':'a', 'khusi':'ki', 'padi':'pi'}

Input = 'push maadi kushi padi'.split()
output = {}
i = 0
while i<len(Input):
    if len(Input[i])%2 == 0:
        output[Input[i]] = Input[i][0] + Input[i][-1]
        
    else:
        output[Input[i]] = Input[i][len(Input[i])//2]
    i+=1
print(output)
```

```python
# Reverse the string without slicing.
s = input('Enter a string :')
i = 0
rev = ''
while i<len(s):
   rev = s[i]+rev
    i+=1
print(rev)
```

```python
# WAP to get the following output:
#  IN = 'abacbaacc'
# OUTPUT = {'a':4, 'b':2, 'c':3}

IN = 'abacbaacc'
i = 0
out = {}
while i<len(IN):
    if IN[i] not in out:
        out[IN[i]] = 1
    else:
        out[IN[i]]+=1
    i+=1
print(out)
```

```python
#  WAP to get the following output:
#  S = ['jiocinema.com', 'file.py', 'web.html', 'amazon.com', 'www.org']
#  output = ['com', 'py', 'html', 'org']

s = ['jiocinema.com', 'file.py', 'web.html', 'amazon.com', 'www.org']
i = 0
out = []
while i<len(s):
    ex = s[i].split('.')[-1]
    if ex not in out:
        out.append(ex)
    i+=1
print(out)
        
```
