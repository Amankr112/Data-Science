#### Intermediate Termination conditions - 
These are the conditions which are used to control the loop.
#### Types :
1) Break
2) Continue
3) Pass

```python
# Break  -  It is a keyword used to terminate the loop by passing some condition.
# Ex - 

i = 1 
n = 10
while i<=10:
    if i == 5:
        break 
    print(i)
    i+=1
```

```python
# Continue  - It is a keyword used to skip the current iteration.

i = 0
while i<=10:
    i+=1          # Increment should be here otherwise when if we will pass increment after continue then after that condition,
    if i == 5:    # the loop will be terminated as after continue it will directly go to the while condition.
        continue
    print(i)

```

```python
# Pass : It is used to make invalid block as valid block.
i = 1
if i>0:
    pass
```

```python
#  WAP to get the following output:
#  S = ['jiocinema.com', 'file.py', 'web.html', 'amazon.com', 'www.org']
#  output = ['com', 'py', 'html', 'org']

s = ['jiocinema.com', 'file.py', 'web.html', 'amazon.com', 'www.org']
i = 0
out = []
while i<len(s):
    ex = s[i].split('.')[-1]
    if ex not in out:
        out.append(ex)
    i+=1
print(out)
        
```
