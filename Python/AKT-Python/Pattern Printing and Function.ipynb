```python
# WAP to find factorial
def factorial(n):
    i = 1
    fact = 1
    while(i<=n):
        fact*= i
        i+=1
    print('Factorial of n :', fact)
factorial(5)
factorial(6)
factorial(7)
```

```python
x = int(input('Enter the number of rows:'))
n = x//2+1
for i in range(n):
    print('  '* (n-i-1), end = '')
    for j in range(2*i+1):
        if j%2 == 0: 
            print(0, end=' ')
        else:
            print(1, end=' ')
    print()

for i in range(n-1):
    print('  '* (i+1), end = '')
    for j in range(2*n-3-2*i):
        if j%2 == 0: 
            print(0, end=' ')
        else:
            print(1, end=' ')
    print()
```
