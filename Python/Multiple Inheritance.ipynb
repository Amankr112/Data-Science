### OOPs - Multiple Inheritance:

```python
class A:
    def method1(self):
        print("A class method is called")
```

```python
class B(A):
    def method1(self):
        print("B class method is called")
    def method2(self):
        print("B class method2 is called")

        
```

```python
class C(A):
    def method1(self):
        print("C class method is called")
```

```python
class D(B,C):
    def method1(self):
        print("D class method is called")
        C.method1(self)
        B.method1(self)
        A.method1(self)
# Here D is inheriting the properties of B, C which are inheriting properties from A. 
```

```python
d=D()
```

```python
d.method1()
```

```python
d.method2()
```

```python
B.method1(d)
```

```python
A.method1(d)
```

#### How Eval works:
1. Parse python Expression
2. Compile into a byte code
3. Evaluate the python expression
4. It will return the result.



```python
var=compile("5*5", "<string>", "eval")
```

```python
var
```

```python
eval(var)
```

#### Globals parameter:

```python
eval("x+50+x**2", {"x":10})
```

```python
x=10
eval("x+50+x**2", {"x":x})
```

```python
x=100
y=12
eval("x+y", {"x":x,"y":y})
```

```python
x=100
y=12
eval("x+y", {"x":x,"y":20})
```

```python
x=100
y=12
eval("x*y+w", {"x":x,"y":y,"w":1000})
```

```python
#Giving Global Parameter
x=100
y=2
w=10
eval("x*y+w")
```

```python
#Giving key value pairs:
eval("x*y+w", {"x":5,"y":8,"w":10})
```

```python
# Giving local Parameter
eval("x * y + w", {"__builtins__": None}, {"x": 5, "y": 8, "w": 10})
#first bracket is global and second one is local
```

```python
a=20
b=12
eval("a>b")
```

```python
#restricting local and global variables 
a=20
b=12
eval("a+b", {}, {})
```

#### Disadvantages of eval():
1. Security Risk – Can execute malicious code if user input is passed.
2. Slow Performance – Slower than regular code; parsed at runtime.
3. Hard to Debug – Errors appear at runtime, not at compile time.
4. Poor Readability – Code inside strings is harder to read and maintain.
5. No Static Analysis – Linters and IDEs can't check inside eval().
6. Risk of Crashes – Can easily raise exceptions (e.g. NameError, SyntaxError).
