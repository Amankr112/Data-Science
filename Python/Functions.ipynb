###  Map function

### List Comprehension
##### List comprehension provide a concise way to create lists. It consists of brackets containing an expression followed by a for clause, then zero or more for or if clauses. The Expression can be anything, meaning you can put in all kind of objects in lists.

```python
lst1=[]
def lst_square(lst):
    for i in lst:
        lst1.append(i*i)
    return lst1
```

```python
lst_square([1,2,3,4,5,6])
```

```python
#List Comprehension
lst=[1,2,3,4,5,6]
[i*i for i in lst]
```

```python
lst1=[i*i for i in lst]
lst1
```

```python
# Square of even numbers in lst:
[i*i for i in lst if i%2==0]
```

```python
# Square of odd numbers in lst:
[i*i for i in lst if i%2!=0]
```

### Python List Iterables vs Iterators

#### Iterables-
- Definition: An object capable of returning its members one at a time.
- Examples: list, tuple, str, set, dict, etc.
- How to get an iterator: Use the built-in iter() function on an iterable.
- Key feature: Can be looped over multiple times (unless modified).

#### Iterators-
- Definition: An object that represents a stream of data, returned one element at a time                using the next() function.
- Created by: Calling iter() on an iterable.
- Key Methods:
##### __iter__() – returns the iterator object itself.
##### __next__() – returns the next item or raises StopIteration.
- Used up after one pass: Cannot be reused after it's exhausted.

```python
# List is Iterable - (An object capable of returning its members one at a time)
lst=[1,2,3,4,5,6,7]
for i in lst:
    print(i)
```

```python
# iter()- used to convert Itrables into iterators
iter(lst)
```

```python
lst1=iter(lst)
```

#### Generator:

```python
def square(n):
    for i in range(n):
        return i**2
```

```python
square(3)
```

```python
def square(n):
    for i in range(n):
        yield i**2
```

```python
square(3)
```

```python
for i in square(3):
    print(i)
```

```python
a=square(3)
```

```python
a
```

```python
next(a)
```

```python
next(a)
```

### Difference:
1. To create iterator we use iter(), and to use generator we use "yield" keyword
2. Generator uses the yield  keyword, It saves the local variable value.
3. Generator in python helps us to write fast and compact code
4.  Iterator is much more memory efficient
