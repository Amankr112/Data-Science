### MatplotLib Tutorial

##### Matplotlib is a plotting library for the Python programming language and its numerical mathematics extension NumPy. It provides an object-oriented API for embedding plots into applications using general-purpose GUI toolkits like Tkinter, wxPython, Qt, or GTK+.

##### Some of the major Pros of Matplotlib are:

- Generally easy to get started for simple plots
- Support for custom labels and texts
- Great control of every element in a figure
- High-quality output in many formats
- Very customizable in general

```python
import numpy as np
```

```python
#Simple Example:
x=np.arange(0,10)
y=np.arange(11,21)
print(x)
print(y)
```

```python
#Scatter:
plt.scatter(x,y,c='y')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Graph in 2D')
plt.savefig('Test.png')
```

```python
plt.plot(x,y, 'r*', linestyle='dashed', linewidth=2, markersize=12)
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('2d Diagram')
```

```python
# Creating Subplots
plt.subplot(2,2,1)
plt.plot(x,y,'b--')

plt.subplot(2,2,2)
plt.plot(x,y,'r')

plt.subplot(2,2,3)
plt.plot(x,y,'g*--')

plt.subplot(2,2,4)
plt.plot(x,y,'yo')

# 1,2,3,4 are indexes
```

```python
x=np.arange(1,11)
y=x**2+5
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.title('Matplotlib demo')
plt.plot(x,y, 'g--')
```

```python
#Convert the x and y coordinates fro points on a sine curve
x=np.arange(0, 4*np.pi, 0.1)
y=np.sin(x)
plt.title("Sine Wave Form")

#Plot the point using matplotlib
plt.plot(x,y)
plt.show()
```

```python
#Subplot()
#Compute the x and y coordinates for the points on sine and cosine curves
x=np.arange(0, 5*np.pi, 0.1)
y_sin=np.sin(x)
y_cos=np.cos(x)

#set up a subplot grid that has height 2 and width 1, and set the first such subplot as active.
plt.subplot(2,1,1)

#First Plot
plt.plot(x, y_sin, 'b--')
plt.title('sine')

#Set the second subplot as active, and make the second plot.
plt.subplot(2,1,2)
plt.plot(x, y_cos, 'r--')
plt.title('cosine')

#show fig.
plt.show()
```

```python
# Bar plot
x=[2,8,10]
y=[11,16,9]

x2=[3,9,11]
y2=[6,15,7]

plt.bar(x,y)
plt.bar(x2,y2, color='r')
plt.title('Bar Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.show()
```

### Histogram

```python
a=np.array([22,87,5,43,56,73,55,54,11,20,51,5,79,31,27])
plt.hist(a)   # here we can alsowrite like - plt.hist(a, bins=20) which means range will be divided into 20 bins
plt.title("Histogram")
plt.show()
#Histogram Plots with respect to count
```

### Box Plot using Matplotlib

```python
data=[np.random.normal(0, std,100) for std in range(1,4)]
#rectangular box plot
plt.boxplot(data, vert=True, patch_artist=True);
#vert=True (means it will show boxplot vertically, if false then show horizontally)
#patch_artist=True (means color wiil be inside the box, if false then colorless)
#o - outlier in the diagram
```

### Pie Chart:

```python
#data to plot:
labels='python', 'C++', 'Java', 'Ruby'
size=[350, 200, 250, 100]
colors=['gold', 'orange', 'skyblue', 'lightgreen']
explode=(0.1, 0, 0, 0)    #explode 1st slice (means 1st slice aur sab slice se kitna dur rahega)

#plot
plt.pie(size, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%', shadow=True)
plt.axis('equal')
plt.show()
```
