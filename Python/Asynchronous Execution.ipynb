#### ThreadPoolExecutor -  
##### It is Asynchronous executor which is preformed with the help of threads.

```python
import time
from concurrent.futures import ThreadPoolExecutor

def returnNumber(a):
    return a
```

```python
returnNumber(12)
```

```python
start=time.time()
with ThreadPoolExecutor(max_workers=10) as executor:
    for result in executor.map(returnNumber, range(50)):
        print("Count: {}".format(result))
print("The total time is :{}".format(time.time()-start))
    
```

##### It will take 1 sec to execute each max_worker(means if max_worker is 5 then after every 5 number the compiler will take 1 sec then execute the next 5 numbers)

```python
def returnNumber(a):
    time.sleep(1)
    return a
```

```python
start=time.time()
with ThreadPoolExecutor(max_workers=5) as executor:
    for result in executor.map(returnNumber, range(50)):
        print("Count: {}".format(result))
print("The total time is :{}".format(time.time()-start))
    
```

##### Two threads are working parallely here, if we remove time.sleep then we will not be able to see the parallel execution because in one take it will print all the numbers.

```python
start=time.time()
with ThreadPoolExecutor(max_workers=10) as executor:
    for result in executor.map(returnNumber, range(50)):
        print("Count: {}".format(result))
print("The total time is :{}".format(time.time()-start))
    
```
