#### SETS:
##### Sets are represented in {}.
##### Sets are iterable, mutable, and has no duplicate elements.
##### Sets always gives values in sorted order (if we are taking numbers).
##### Sets not support Indexing.

```python
set_var=set()
print(set_var)
print(type(set_var))
```

```python
set_var={'Avengers', 'IronMan', 'Krish'}
print(set_var)
type(set_var)
```

```python
# Gives Sorted set
set_var={1,2,4,2,6,3,5}
print(set_var)
```

```python
#Inbuilt functions in set
# 1. add() - To add elements
set_var.add(7)
set_var
```

```python
set1={'Avengers', 'IronMan', 'Hitman'}
set2={'Avengers', 'IronMan', 'Hitman', 'Hulk'}
```

```python
set2.intersection(set1)
```

```python
# Difference - common elements will be removed
set2.difference(set1)   #from set2 remove the elements of set1
```

```python
set2
```

```python
#Difference_Update
set2.difference_update(set1)    #update will change the set 
```

```python
set2
```

##### Dictionarie is collection which is Unorderd, changable, and indexed

```python
# Dictionary Creation:
my_dict={1:"Aman", 2:"Kumar", 3:"Tiwari"}
my_dict
```

```python
type(dict())
```

```python
my_dict[1]
```

```python
# We can have loop throught the dictionaries keys
for x in my_dict:
    print(x)
```

```python
# We can have loop throught the dictionaries values
for x in my_dict.values():
    print(x)
```

```python
# We can have also check both keys and values
for x in my_dict.items():
    print(x)
```

#### Nested Dictionary

```python
car1_model={'Marcedes': 1960}
car2_model={'BMW': 1970}
car3_model={'Audi': 1980}
car_type={'car1':car1_model, 'car2':car2_model, 'car31':car3_model}
```

```python
car_type
```

```python
# Accessing the items in the dictionary
print(car_type['car1'])
```

```python
print(car_type['car1']['Marcedes'])
```
