### Data Classes in Python

**Data Classes** in Python are a feature (introduced in **Python 3.7**) that provide a simple way to create classes used mainly for storing data, without writing a lot of boilerplate code like `__init__`, `__repr__`, and `__eq__`.

They are created using the `@dataclass` decorator from the `dataclasses` module.

---

#### Key Features
- Automatically generates `__init__()`, `__repr__()`, and `__eq__()` methods.
- Reduces boilerplate code.
- Supports **default values** and **type hints**.
- Immutable data classes using `frozen=True`.


#### Advantages of Data Classes:
- Less code, more readability.
- Built-in comparison support.
- Great for structured data (like configs or records).
- Works well with type hints.


```python
from dataclasses import dataclass

@dataclass 
class Person:
    name:str
    age:int
    profession:str
person1=Person("AMAN", 22, "MLOPS")
print(person1)
```

```python
person1.profession="AI Engineer"
```

```python
print(person1)
```

```python
# Immutable class - Uses (frozen=True) which makes the variable static.
@dataclass(frozen=True)
class Point:
    x:int
    y:int

point=Point(4, 5)
print(point)
```

```python
point.x = 35
```

### Inheritance -

```python
@dataclass
class Person:
    name: str
    age: int

@dataclass
class Employee(Person):
    employee_id: int
    department: str

person=Person("Aman", 22)
employee=Employee("Aman", 22, 2345, "CSE")
```

```python
print(employee)
```

### Nested Dataclass

```python
@dataclass 
class Address: 
    street : str
    city : str
    pin_code : int

@dataclass
class Person:
    name : str
    age : int
    address : Address
address=Address("1 Bangla Chowk pitiz", "Chatra", 825408)
person=Person("Aman", 22, address)
```

```python
person
```

```python
person.address.street
```
