```python
lst1=['Aman', 'Om', 'Sneha']
lst2=['a', 'b', 'c']
```

```python
output=zip(lst1, lst2)
for i, j in output:
    print(i, j)
```

```python
lst1=['Aman', 'Om', 'Sneha']
lst2=['a', 'b', 'c']
lst3=[1, 2, 3]
```

```python
output=zip(lst1, lst2, lst3)
for i, j, k in output:
    print(i, j, k)
```

```python
lst1=['Aman', 'Om', 'Sneha', 4]  
lst2=['a', 'b', 'c']
lst3=[1, 2, 3]  
```

```python
output=zip(lst1, lst2, lst3)
for i, j, k in output:
    print(i, j, k)
```
