### Exception handling in Python

####  Error:
- Definition: Errors are problems in a program that cannot be handled or recovered from during execution.
- Example: Syntax errors, memory errors.
- Occurs When: The code itself is incorrect or ill-formed.
- Handled by?: Usually not handled with try/except.

#### Exception:
- Definition: Exceptions are events that can be handled during program execution to prevent crashes.
- Example: ZeroDivisionError, FileNotFoundError, IndexError, etc.
- Occurs When: The code is syntactically correct but encounters an issue at runtime.
- Handled by?: Yes, using try...except blocks.

```python
try:
    # Code block where exception can occur
    a=b
except:
    print("Some problem may have occured")
```

```python
#Using try-else
try:
    a=int(input("Enter a:"))
    b=int(input("Enter b:"))
    c=a+b
    d=a*b
    e=a/b
    f=a%b
    
except NameError:   
    print("The user have not defined the variable")
except TypeError:   
    print("Try to make the datatype similar")
except ZeroDivisionError:
    print("Division from 0 is not allowed")

except Exception as ex:
    print(ex)

else:
    print(c)
    print(d)
    print(e)
    print(f)
```

```python
#try-else-finally

try:
    a=int(input("Enter a:"))
    b=int(input("Enter b:"))
    c=a+b
    
except NameError:   
    print("The user have not defined the variable")
except TypeError:   
    print("Try to make the datatype similar")
except ZeroDivisionError:
    print("Division from 0 is not allowed")

except Exception as ex:
    print(ex)

else:       #if their is no exception then else part willl be executed
    print(c)
finally:      #finally block always get printed for (cleanup, closing file)
    print("The Execution is done")
```

### Custom Exception:

```python
class Error(Exception):
    pass
class dobException(Error):
    pass
```

```python
year=int(input("Enter the year of birth:"))
age=2025-year

try:
    if age<=30 and age>20:
        print("The age is valid. You can apply for the exam")
    else:
        raise dobException
except dobException:
    print("The age is not within the range")
        
    
```

### Python Assert
##### Python provides the assert statement to check if a given logical expression is true or false. Program execution proceeds only if the expression is true and raises the AssertionError when it is false. The following code shows the usage of the assert statement.


```python
10>10
```

```python
10>=10
```

```python
num=10
assert num>=10
```

```python
num=10
assert num>10   # Raises AssertionError  
```

```python
try:
    num=int(input("Enter the number"))
    assert num%2==0
    print("The number is even")
except AssertionError:
    print("Please enter even number")
```

```python
try:
    num=int(input("Enter the number"))
    assert num%2==0
    print("The number is even")
except AssertionError:
    print("Please enter even number")
```
