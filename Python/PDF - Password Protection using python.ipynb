```python
from PyPDF2 import PdfReader, PdfWriter
```

```python
# Open the current pdf:
file_pdf=PdfReader('Knowledge.pdf')
# Object for pdf writer
out_pdf=PdfWriter()
```

```python
file_pdf
```

```python
print(len(file_pdf.pages))
```

```python
for i in range(len(file_pdf.pages)):
    page_details = file_pdf.pages[i]  # Access page directly
    out_pdf.add_page(page_details)    # Add page to writer
```

```python
password="Aman@123"
out_pdf.encrypt(password)
```

```python
with open("Encrypted_Knowledge.pdf", "wb") as filename:
    out_pdf.write(filename)
    
```
