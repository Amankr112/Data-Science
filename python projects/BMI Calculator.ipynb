```python
Height = float(input("Enter your height in centimeters: "))
Weight = float(input("Enter your weight in Kg: "))

Height = Height / 100  # Converts cm to meters
BMI = Weight / (Height * Height)

print("Your Body Mass Index is:", BMI)

if BMI > 0:
    if BMI <= 16:
        print("You are severely underweight")
    elif BMI <= 18.5:
        print("You are underweight")
    elif BMI <= 25:
        print("You are Healthy")
    elif BMI <= 30:
        print("You are overweight")
    else:
        print("You are severely overweight")
else:
    print("Invalid details. Please enter valid details.")
```

```python
# Upadted BMI Calculator
def calculate_BMI():
    try:
        height = float(input("Enter your height in centimeters: "))
        weight = float(input("Enter your weight in kilograms: "))

        if(height<=0 or weight<=0):
            print("\033[91mError: Height and weight must be positive numbers.\033[0m")
            return

        height=height/100
        BMI=weight/(height*height)
        BMI=round(BMI,2)    #The result is rounded to 2 decimal places using round().
        
        print(f"\n\033[96mYour Body Mass Index (BMI) is: {BMI}\033[0m")
        
        # BMI Category and Color-coded Message
        if BMI <= 16:
            print("\033[91mCategory: Severely Underweight\033[0m")
        elif BMI <= 18.5:
            print("\033[93mCategory: Underweight\033[0m")
        elif BMI <= 25:
            print("\033[92mCategory: Healthy\033[0m")
        elif BMI <= 30:
            print("\033[93mCategory: Overweight\033[0m")
        else:
            print("\033[91mCategory: Severely Overweight\033[0m")

        # Display reference chart
        print("\n\033[95mBMI Classification Chart:\033[0m")
        print("""
        Less than 16         : Severely Underweight
        16 – 18.5            : Underweight
        18.5 – 25            : Healthy
        25 – 30              : Overweight
        More than 30         : Severely Overweight """)

    except ValueError:
        print("\033[91mInvalid input! Please enter numeric values only.\033[0m")
```

```python
calculate_BMI()
```
