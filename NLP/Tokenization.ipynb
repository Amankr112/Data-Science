### 2. Regular Expression :

```python
import re 
sent3 = 'I am going to delhi!'
tokens = re.findall("[\w']+" , sent3) 
tokens
```

### 3.NLTK :

##### punkt → Main sentence tokenizer models.

##### punkt_tab → Tokenizer tables introduced in newer NLTK versions (extra data needed for sent_tokenize).

```python
import nltk
nltk.download('punkt')
nltk.download("punkt_tab")   
```

```python
from nltk.tokenize import sent_tokenize, word_tokenize  

# Sentence Tokanization
text = "Dr. Smith is a good teacher. He teaches at U.S.A. University. Students like him!"
sents = sent_tokenize(text)
```

```python
sents
```

```python
 # word_tokenization
for sent in sents:   
    print(word_tokenize(sent))   
```

```python
sent5 = "I have a Ph.D in A.I"
sent6 = "We're here to help! mail us at nks@gmail.com"
sent7 = "A 5km ride cost $10.50"
```

```python
word_tokenize(sent5)
```

### 4. Spacy (Best NLP library)

```python
import spacy
nlp = spacy.load('en_core_web_sm')
```

```python
# we have to convert the sentences into documents
doc1 = nlp(sent5)
doc2 = nlp(sent6)
doc3 = nlp(sent7)
doc4 = nlp(sent1)
```

```python
for token in doc4:
    print(token)
```
