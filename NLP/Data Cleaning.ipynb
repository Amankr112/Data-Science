### Removing HTML tags :

```python
sample_text = '<p id="removable-paragraph">This is a sample paragraph that you can remove later.</p>'
```

```python
sample_text
```

```python
import re  # remove

def striphtml(data):
    p = re.compile(r'<.*?>')
    return p.sub('', data)
```

```python
striphtml(sample_text)
```
