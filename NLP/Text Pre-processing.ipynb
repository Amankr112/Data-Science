### Removing URLs :

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.S+')
    return pattern.sub(r'', text)
```

### Removing Punctuations

```python
import string
string.punctuation
```

```python
exclude = string.punctuation
```

```python
def remove_punctuation(text):
    for char in exclude:
        text = text.replace(char, '')
    return text
```

```python
text = 'string. with. punctuation?'
```

```python
print(remove_punctuation(text))
```

##### The function we are using above to remove punctuations are build logically, and for larger dataset it will take more time

```python
import time
start = time.time()
print(remove_punctuation(text))
time1 = time.time() - start
print(time1*50000)
```

#### The standard function for Removing Punctuations :

```python
def remove_punc(text):
    return text.translate(str.maketrans('', '', exclude))
```

```python
start = time.time()
print(remove_punc(text))
time2 = time.time() - start
print(time2*50000)
```

```python
def chat_conversion(text):
    new_text = []
    for x in text.split():
        if x.upper() in chat_words:
            new_text.append(chat_words[x.upper()])
        else:
            new_text.append(x)
    return " ".join(new_text)
```

### Removing Stop words :  ( a, an, the, are, my, ......)

```python
from nltk.corpus import stopwords
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text = []

    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')

        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)
```

```python
remove_stopwords('Despite the heavy rain and the unusually strong winds, the students continued their outdoor science experiment with great enthusiasm, carefully recording every observation in their notebooks so that they could later analyze the data and prepare a detailed report for their upcoming class presentation.')
```

### Emojis Handling:

```python
import re

def remove_emoji(text):
    emoji_pattern = re.compile(
        "["
        u"\U0001F600-\U0001F64F"  # emoticons
        u"\U0001F300-\U0001F5FF"  # symbols & pictographs
        u"\U0001F680-\U0001F6FF"  # transport & map symbols
        u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
        u"\U00002702-\U000027B0"  # dingbats
        u"\U000024C2-\U0001F251"  # enclosed characters
        "]+",
        flags=re.UNICODE
    )
    return emoji_pattern.sub(r'', text)



remove_emoji ("Hello 😊🚀🌍! Let's remove emojis 👍🔥.")
```

### Stemming :
##### Stemming is the process of reduing inflection in words to root forms such as mapping a group of words to the same stem even if the stem itself is not a valid word in the language.

- Stemming is a text preprocessing technique in NLP where words are reduced to their root/base form by chopping off prefixes or suffixes. It’s a rule-based, crude process (not always linguistically correct) but useful for search engines, indexing, and IR tasks.

```python
from nltk.stem import PorterStemmer
```

```python
ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
sample = 'walk walks walking walked'
stem_words(sample)
```

### Lemmatization :
##### Lemmatizatio, unlike Stemming, reduces the inflected words properly ensuring that the root word belongs to the language. In Lemmatization root word is called Lemma. A lemma is the canonical form , dictionary form, or citation form of a set of words.

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer

wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations = "?:!.',"

sentence_words = nltk.word_tokenize(sentence)

for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

print(f"{'Word':<20}{'Lemma'}")
for word in sentence_words:
    print(f"{word:<20}{wordnet_lemmatizer.lemmatize(word)}")
```
